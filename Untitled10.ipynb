```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


faq_data = {
    "What is CodeAlpha?": "CodeAlpha is a platform offering virtual internships to students to enhance their technical skills.",
    "How many tasks should I complete?": "You need to complete at least 2 or 3 tasks from the given task list to get the certificate.",
    "Where do I submit my tasks?": "You need to submit your projects via the official Submission Form shared in your WhatsApp group.",
    "Will I get a Letter of Recommendation?": "Yes, a Letter of Recommendation (LOR) will be provided based on your project performance.",
    "What is the deadline for submission?": "Please check the timeline details provided in your official internship offer letter or WhatsApp group.",
}

questions = list(faq_data.keys())
answers = list(faq_data.values())


def get_bot_response(user_query):
    if not user_query.strip():
        return "Please ask something!"

    all_questions = questions + [user_query]

    vectorizer = TfidfVectorizer(stop_words="english")
    tfidf_matrix = vectorizer.fit_transform(all_questions)


    similarity_scores = cosine_similarity(tfidf_matrix[-1], tfidf_matrix[:-1])

    best_match_idx = similarity_scores.argmax()
    highest_score = similarity_scores[0][best_match_idx]


    if highest_score > 0.3:
        return answers[best_match_idx]
    else:
        return "Sorry, I couldn't find an exact match for that. Please try rephrasing your question!"


# 3. Chat Application Loop
print(
    "🤖 Bot: Hello! I am your CodeAlpha Assistant. (Type 'exit' to stop)\n"
)
print("-" * 50)

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        print("🤖 Bot: Goodbye! Happy learning!")
        break

    reply = get_bot_response(user_input)
    print(f"🤖 Bot: {reply}")
    print("-" * 50)
```
